```python
import pandas as pd
```

## Criar um DataFrame para cada ano

```python
df_2016 =  pd.read_csv("compras-e-licitacoes-2016.csv")
```

```python
df_2017 =  pd.read_csv("compras-e-licitacoes-2017.csv")
```

```python
df_2018 = pd.read_csv("compras-e-licitacoes-2018.csv")
```

```python
df_2019 = pd.read_csv("compras-e-licitacoes-2019.csv")
```

```python
df_2020 = pd.read_csv("compras-e-licitacoes-2020.csv")
```

## Criar um DataFrame com todos os anos

```python
total_df = pd.concat([df_2016, df_2017, df_2018, df_2019, df_2020])
total_df.info()
```

## Criar uma copia do DataFrame Original

```python
total__bck_df = total_df.copy()
```

## Converter as Datas que estão como Object para DateTime
	Dia Abert Sessão Publ ID
	Dia Homologação Compra DESC
	Dia Homologação Compra ID

```python
for colunas_data in [
    'Dia Abert Sessão Publ ID',
    'Dia Homologação Compra DESC',
    'Dia Homologação Compra ID',
]:
    total_df[colunas_data] = pd.to_datetime(total_df[colunas_data], errors='coerce')
    total_df.info()
```

## Converter os numeros que estão como Object para to_number
	Valor Total Homologado

```python
from numpy import e
# Removendo ponto milhar
total_df["Valor Total Homologado"] = total_df["Valor Total Homologado"].astype(str).str.replace('.', '', regex=False)

# Trocando vírgula decimal por ponto
total_df["Valor Total Homologado"] = total_df["Valor Total Homologado"].str.replace(',', '.', regex=False)

# Convertando object em float64
total_df["Valor Total Homologado"] = pd.to_numeric(total_df["Valor Total Homologado"], errors='coerce')
total_df.info()
```

## Converter o float64 para Inter64
	Ano Resultado Compra DESC

```python
# Transforma o conteúdo da coluna em números, sem erro
total_df["Ano Resultado Compra DESC"] = pd.to_numeric(total_df["Ano Resultado Compra DESC"], errors='coerce')

# Transforma o conteúdo da coluna de float64 em Int64
total_df["Ano Resultado Compra DESC"] = total_df["Ano Resultado Compra DESC"].astype('Int64')
total_df.info()
```

## Criando uma coluna "Ano" para verificar os gastos anuais
	Dia Homologação Compra DESC

```python
# Criar uma nova coluna no DataFrame somente com o Ano  do dia da homologação
homologacao_ano = "Dia Homologação Compra DESC"

# Transformando a data em um número que pode ser usado como chave de agrupamento.
total_df[homologacao_ano] = pd.to_datetime(total_df[homologacao_ano], errors='coerce', dayfirst=True)

# Garantindo que o número seja inteiro (2016 - 2020)
total_df['Ano'] = total_df[homologacao_ano].dt.year.astype('Int64')
total_df.info()
```

## Retirando os valores nulos da coluna "Ano"

```python
total_df.dropna(subset=['Ano'], inplace=True)
total_df.info()
```

```python
total_df["Ano"]
```

## Eliminado anos anterios a 2016


```python
# Definindo as condições
ano_min = total_df["Ano"] >= 2016
ano_max = total_df["Ano"] <= 2020

# Combinando as condições (E lógico, representado pelo &)
ano_valido = ano_min & ano_max
```

```python
total_df = total_df[
    (total_df["Ano"] >= 2016) &
    (total_df["Ano"] <= 2020)
]
```

```python
total_df.info()
total_df["Ano"]
```

## Gastos por ano

```python
gastos_por_ano = total_df.groupby("Ano")["Valor Total Homologado"].sum()
pd.options.display.float_format = 'R$ {:,.2f}'.format
display(gastos_por_ano)
```

## Gastos por Município

```python
gastos_por_municipio = total_df.groupby("Municipio UResp Compra DESC")["Valor Total Homologado"].sum()
display(gastos_por_municipio)
```

## Gastos por Município por Ano

```python
gastos_municipio_ano = total_df.groupby(["Municipio UResp Compra DESC", "Ano"])["Valor Total Homologado"].sum()
display(gastos_municipio_ano)
```

## Maior Gasto

```python
indice_maximo = total_df["Valor Total Homologado"].idxmax()

registro_maximo = total_df.loc[indice_maximo]

resultado = registro_maximo[["Municipio UResp Compra DESC", "Ano", "Valor Total Homologado"]]
display(resultado)
```

## Menor Gasto

```python
indice_minimo = total_df["Valor Total Homologado"].idxmin()

registro_minimo = total_df.loc[indice_minimo]
resultado = registro_minimo[["Municipio UResp Compra DESC", "Ano", "Valor Total Homologado"]]
display(resultado)
```
