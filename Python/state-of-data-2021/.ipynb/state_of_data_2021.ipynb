```python
import pandas as pd
```

```python
state_df = pd.read_excel("state-of-data-2021.xlsx")
state_df.info()
```

```python
state_df_bck = state_df.copy()
```

```python
## Excluindo a coluna ID
```

```python
state_df.drop("ID", axis=1, inplace=True)
state_df.info()
```

## Alterando a coluna ["Idade"] de float64 par Int64

```python
state_df["Idade"] = pd.to_numeric(state_df["Idade"], errors="coerce")

state_df["Idade"] = state_df["Idade"].astype('Int64')
state_df.info()
```

## Preenchendo a coluna ["Regiao de origem"]

```python
state_df["Regiao de origem"] = state_df["Regiao de origem"].fillna(
    state_df["Regiao onde mora"]
)
state_df["Regiao de origem"]
```

## Preemchendo a coluna ["Mudou de Estado ?"]

```python
import numpy as np
```

```python
condicao_mudanca = state_df["Regiao onde mora"] != state_df["Regiao de origem"]
state_df["Mudou de Estado ?"] = np.where(
    condicao_mudanca,
    'Sim',
    'Não'
)
state_df["Mudou de Estado ?"]
```

## Verificando os valores nulos da coluna ["Regiao de origem"]

```python
contagem_nulos = state_df["Regiao de origem"].isnull().sum()

print(f"Total de valores nulos em 'Regiao de origem': {contagem_nulos}")
```

```python
state_df["Regiao de origem"] = state_df["Regiao de origem"].fillna(
    state_df["Regiao onde mora"]
)
```

## Excluindo valores nulos das colunas ["Idade", "Faixa salarial"]

```python
valores_nulos = ["Idade", "Faixa salarial"]
print(valores_nulos)
```

```python
state_df = state_df.dropna(subset=valores_nulos)
state_df.info()
```

## Salário por faixa etária

```python
salario_por_faixa_etaria = state_df.groupby(["Faixa idade", "Faixa salarial"]).size().unstack(fill_value=0)
print(salario_por_faixa_etaria)
```

## Salário por região

```python
salario_por_regiao = state_df.groupby(["Regiao onde mora", "Faixa salarial"]).size().unstack(fill_value=0)
print(salario_por_regiao)
```

## Salário por gênero

```python
salario_por_genero = state_df.groupby(["Genero", "Faixa salarial"]).size().unstack(fill_value=0)
print(salario_por_genero)
```

## Salário por região do gênero masculino

```python
df_masculino = state_df[state_df["Genero"] == "Masculino"]

salario_regiao_masc_abs = df_masculino.groupby(
    ["Regiao onde mora", "Faixa salarial"]
).size().unstack(fill_value=0)
print(salario_regiao_masc_abs)
```

## Salário por região do gênero feminio

```python
df_feminino = state_df[state_df["Genero"] == "Feminino"]

salario_regiao_fem_abs = df_feminino.groupby(
    ["Regiao onde mora", "Faixa salarial"]
).size().unstack(fill_value=0)
print(salario_regiao_fem_abs)
```
